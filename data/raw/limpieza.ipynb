```python
import os
import pandas as pd

# Verificar el archivo en la ruta directa o subiendo un nivel
nombre_archivo = 'normales_climatologicas_estandar_periodo_1991-2020.xlsx'

if os.path.exists(f'raw/{nombre_archivo}'):
  ruta_raw = f'raw/{nombre_archivo}'
elif os.path.exists(f'data/raw/{nombre_archivo}'):
  ruta_raw = f'data/raw/{nombre_archivo}'
else:
  ruta_raw = nombre_archivo

# Cargar el archivo Excel oficial del IDEAM
df_raw = pd.read_excel(ruta_raw, sheet_name='PRECIPITACIÓN', skiprows=1)

print(f'Registros originales en la base de datos: {len(df_raw)}')
df_raw.head()
```

```python
# 1. Seleccionar las columnas principales
columnas_interes = [
    'CÓDIGO',
    'CATEGORÍA',
    'ESTACIÓN',
    'MUNICIPIO',
    'DEPARTAMENTO',
    'ALTITUD (m)',
    'LATITUD',
    'LONGITUD',
    'ENE',
    'FEB',
    'MAR',
    'ABR',
    'MAY',
    'JUN',
    'JUL',
    'AGO',
    'SEP',
    'OCT',
    'NOV',
    'DIC',
    'ANUAL',
]
df_clean = df_raw[columnas_interes].copy()

# 2. Renombrar columnas a formato estándar
df_clean.columns = [
    'codigo',
    'categoria',
    'estacion',
    'municipio',
    'departamento',
    'altitud_m',
    'latitud',
    'longitud',
    'ene',
    'feb',
    'mar',
    'abr',
    'may',
    'jun',
    'jul',
    'ago',
    'sep',
    'oct',
    'nov',
    'dic',
    'precipitacion_anual_mm',
]

# 3. Eliminar duplicados y registros sin municipio o precipitación anual
df_clean = df_clean.drop_duplicates(subset=['codigo'])
df_clean = df_clean.dropna(subset=['municipio', 'precipitacion_anual_mm'])

# 4. Redondear valores numéricos
columnas_meses = [
    'ene',
    'feb',
    'mar',
    'abr',
    'may',
    'jun',
    'jul',
    'ago',
    'sep',
    'oct',
    'nov',
    'dic',
    'precipitacion_anual_mm',
]
df_clean[columnas_meses] = df_clean[columnas_meses].round(1)

print(f'Registros limpios finales: {len(df_clean)}')
```

```python
# Exportar CSV procesado que usará la aplicación Streamlit
df_clean.to_csv('datos.csv', index=False)
print('¡Archivo data/datos.csv creado exitosamente!')
```
